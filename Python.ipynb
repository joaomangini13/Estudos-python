```python
  def cadastrar_aluno():
    # 1. Pede o nome do aluno
    nome = input("Digite o nome do aluno: ").strip()
    notas = []

    print(f"\n--- Cadastro de notas para {nome} ---")
    print("(Digite 'fim' para encerrar a inserção de notas)\n")

    # 2. Loop para pedir quantas notas o usuário quiser
    while True:
        entrada = input("Digite uma nota (ou 'fim' para encerrar): ").strip()

        # Verifica se o usuário quer parar
        if entrada.lower() == 'fim':
            break

        # Tenta converter a entrada para um número decimal (float)
        try:
            nota = float(entrada)
            if 0 <= nota <= 10:  # Garante que a nota está entre 0 e 10
                notas.append(nota)
                print(f" Nota {nota} adicionada com sucesso!")
            else:
                print(" A nota deve ser um valor entre 0 e 10.")
        except ValueError:
            print(" Entrada inválida! Digite um número ou 'fim'.")

    # 3. Monta o dicionário (como a linha de um banco de dados)
    aluno = {
        "nome": nome,
        "notas": notas
    }

    return aluno


def analisar_aluno(dados):
    # Se a lista de notas estiver vazia
    if len(dados["notas"]) == 0:
        return f"\nO aluno {dados['nome']} não possui notas cadastradas."

    media = sum(dados["notas"]) / len(dados["notas"])

    if media >= 7.0:
        status = "Aprovado"
    else:
        status = "Recuperação"

    return f"\nAluno: {dados['nome']} | Média: {media:.2f} | Status: {status}"


# --- EXECUTANDO O PROGRAMA ---
# Passo A: Cadastra o aluno digitando na tela
novo_aluno = cadastrar_aluno()

# Passo B: Passa o dicionário criado para a função de análise
resultado = analisar_aluno(novo_aluno)
print(resultado)
```
